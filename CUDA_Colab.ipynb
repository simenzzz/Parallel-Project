```python
# 4. Automatically Download Results
import glob
import os
from google.colab import files

try:
    # Find the latest results directory
    latest_run = max(glob.glob('results/run_*'), key=os.path.getmtime)
    cuda_csv = os.path.join(latest_run, 'cuda.csv')
    
    if os.path.exists(cuda_csv):
        print(f"Found results at {cuda_csv}. Downloading...")
        files.download(cuda_csv)
    else:
        print(f"Error: {cuda_csv} not found.")
except ValueError:
    print("Error: No results directories found in results/")
```
